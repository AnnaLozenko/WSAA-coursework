## Assignment 4: GitHub

Write a program in python that will read a file from a repository. The program should then replace all the instances of the text "Andrew" with your name.

The program should then commit those changes and push the file back to the repository (You will need authorisation to do this).

```python
# import requests for making HTTP requests
import requests
# import github for interacting with the GitHub API, Auth for authentication
from github import Github, Auth
# import the API keys from the configuration file
from config import apikeys as cfg
```

I will use the PyGithub library to interact with the GitHub API. First, I will authenticate using my GitHub API key. The API key is stored in a separate configuration file for security reasons.

```python
# authenticate with GitHub using the API key
apikey = cfg["githubkey"]
auth = Auth.Token(apikey)
g = Github(auth=auth)
```

```python
# Load repository information
repo = g.get_repo("AnnaLozenko/aprivateone")
```

```python
# get the download URL for the file I want to modify
fileInfo = repo.get_contents("assignment_4_text_file.txt")
url = fileInfo.download_url
```

```python
# check the status code of the response to ensure the file was retrieved successfully and print the content of the file
response = requests.get(url)
if response.status_code == 200:
    file_content = response.text
    print(file_content)
else:
    print(f"Failed to retrieve the file. Status code: {response.status_code}")
```

```python
# replace all instances of "Andrew" with "Anna" in the file content
new_content = file_content.replace("Andrew", "Anna")
print(new_content)
```

```python
# get the SHA of the file, which is essential to update the file in the repository
sha = fileInfo.sha
path = fileInfo.path
```

```python
# update the file in the repository with the new content and commit the changes
new_response = repo.update_file(path, "Update file", new_content, sha, branch="main")
if new_response: # check if the update was successful
    print("File updated successfully.")
    print(response.status_code)
else:    print("Failed to update the file.")
```
